```python
import os
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = "GOOGLE_API_KEY"
GROQ_API_KEY = "GROQ_API_KEY"
OPENAI_API_KEY = "OPENAI_API_KEY"

def require_env(name: str) -> str:
    value = os.getenv(name)
    if value is None:
        raise ValueError(f"{name} is not set. Add it to your .env file.")
    os.environ[name] = value
    return value

google_api_key = require_env(GOOGLE_API_KEY)
groq_api_key = require_env(GROQ_API_KEY)
openai_api_key = require_env(OPENAI_API_KEY)
```

## Streaming and Batch

Some models can stream their content while it is being generated.

```python
# It returns a generator object
model.stream("Write me a 100 words paragraph on A.I.")
```

```python
for chuck in model.stream("Write me a 100 words paragraph on A.I."):
    print(chuck.text, end="|", flush=True)
```

```python
# Άλλαξα το chuck σε chunk για σωστή ορθογραφία
for chunk in model.stream("Write me a 100 words paragraph on A.I."):
    # Χρησιμοποιούμε chunk.content και end="" για συνεχή ροή
    print(chunk.content, end="", flush=True)
```

### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
# Batch

responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response)
```

```python
model.batch(
    [
        "Why do parrots have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?"
    ],
    max_concurrency=5   # Limit to 5 parallel calls
)
```
